```python
from transformers import SamModel, SamProcessor, DetrImageProcessor, DetrForObjectDetection
from tqdm import tqdm
from src.Preprocessor.core import process_one_image
```

```python
detector_processor = DetrImageProcessor.from_pretrained("facebook/detr-resnet-50")
detector = DetrForObjectDetection.from_pretrained("facebook/detr-resnet-50", revision="no_timm")

sam_processor = SamProcessor.from_pretrained("facebook/sam-vit-base")
sam_model = SamModel.from_pretrained("facebook/sam-vit-base")

fail_count = 0
image_files = ["../input/imgs/train/c1/img_100021.jpg",
               "../input/imgs/train/c2/img_100113.jpg", "../input/imgs/train/c3/img_100281.jpg",
               "../input/imgs/train/c4/img_100343.jpg", "../input/imgs/train/c6/img_100540.jpg",
               "../input/imgs/train/c7/img_100285.jpg", "../input/imgs/train/c8/img_100467.jpg,",
               "../input/imgs/train/c9/img_100299.jpg"]

for f in tqdm(image_files, desc="Processing images"):
    try:
        result = process_one_image(
            f,
            "./out",
            detector,
            detector_processor,
            sam_model,
            sam_processor,
            "cpu",
            smooth_k=5,
            feather_k=25,
            bg_blur_k=15,
            bg_darken_factor=0.5,
        )
        if not result["ok"]:
            print(f"Failed to process {f}: {result['error']}")
            fail_count += 1
    except Exception as e:
        print(f"Critical error on {f}: {e}")
        fail_count += 1
```
